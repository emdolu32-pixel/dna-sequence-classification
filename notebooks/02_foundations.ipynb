```python
import pandas as pd

df = pd.DataFrame({
    "sequence": ["ATGCGC", "ATATATTA", "GGCCGGCCAA", "ATGAAATTT", "CGCGAT"],
    "family":   ["kinase", "receptor", "kinase", "receptor", "channel"],
})
df
```

```python
print(df.shape)
print(df.head(2))
print(df.family)
```

```python
df["length"] = df["sequence"].str.len()
df
```

```python
def gc_content(seq):
    return (seq.count("G") + seq.count("C")) / len(seq)

df["gc"] = df["sequence"].apply(gc_content)
df
```

```python
df["family"].value_counts()
```

```python
df[df["gc"]>0.5]
```
